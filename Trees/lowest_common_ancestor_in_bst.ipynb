Given a binary search tree (BST) where all node values are unique, and two nodes from the tree p and q, return the lowest common ancestor (LCA) of the two nodes.

The lowest common ancestor between two nodes p and q is the lowest node in a tree T such that both p and q as descendants. The ancestor is allowed to be a descendant of itself.

Example 1:



Input: root = [5,3,8,1,4,7,9,null,2], p = 3, q = 8

Output: 5
Example 2:



Input: root = [5,3,8,1,4,7,9,null,2], p = 3, q = 4

Output: 3
Explanation: The LCA of nodes 3 and 4 is 3, since a node can be a descendant of itself.

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

class Solution:
    def lowestCommonAncestor(self, root: TreeNode, p: TreeNode, q: TreeNode) -> TreeNode:
        #NB This is a BST hence lefts of every node are always less than rights

        curr = root                 #start from root
        while curr:                 
            if p.val < curr.val and q.val < curr.val:   
                curr = curr.left

            elif p.val > curr.val and q.val > curr.val:
                curr = curr.right

            else:                       #if any of p or q = curr.val then return it. Thats out LCA
                return curr

#NB: the last else captures these scenarios:
# All three cases mean `curr` is the LCA:

# **Case 1 & 2:** If p or q equals curr, then curr itself is the LCA (a node is an ancestor of itself)

# **Case 3:** If p is on the left and q is on the right (or vice versa), then curr is the first node that "splits" them apart — making it their lowest common ancestor.

# Time Complexity: O(h) where h is the height of the tree. In worst case, it can be O(n) for skewed tree.
# Space Complexity: O(1) since we are using constant space.
```
