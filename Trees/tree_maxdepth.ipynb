Given the root of a binary tree, return its depth.

The depth of a binary tree is defined as the number of nodes along the longest path from the root node down to the farthest leaf node.

Example 1:



Input: root = [1,2,3,null,null,4]

Output: 3
Example 2:

Input: root = []

Output: 0
Constraints:

0 <= The number of nodes in the tree <= 100.
-100 <= Node.val <= 100

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

class Solution:
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        #catch edge cases
        if root is None:
            return 0

        #use recurssion to capture left and right depth
        left_depth = self.maxDepth(root.left)
        right_depth = self.maxDepth(root.right)

        #return the max depth between the left and right. Add the root as well ie 1
        return max(left_depth, right_depth) + 1
#TC: O(n) - we visit every node
#SC: O(h) - the maximum depth of the tree, due to recursion stack
```
