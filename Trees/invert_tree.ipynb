```python
# Invert Tree
def invertTree(root):
    # Step 1: Edge case - empty tree returns None
    if not root:
        return None
    
    # Step 2: Trust the algorithm - invert both children
    left_inverted = invertTree(root.left)
    right_inverted = invertTree(root.right)
    
    # Step 3: Combine results - swap the inverted children
    root.left = right_inverted
    root.right = left_inverted
    return root

#TC: O(n) - we visit each node once
#SC: O(h) - the maximum depth of the tree, due to recursion stack
```
