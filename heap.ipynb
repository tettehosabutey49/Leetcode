Given an integer array nums and an integer k, return the kth largest element in the array.

Note that it is the kth largest element in the sorted order, not the kth distinct element.

Can you solve it without sorting?

 

Example 1:

Input: nums = [3,2,1,5,6,4], k = 2
Output: 5
Example 2:

Input: nums = [3,2,3,1,2,4,5,5,6], k = 4
Output: 4
 

```python
#import heapq

class Solution(object):
    def findKthLargest(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        # nums.sort()     #TC: O(nlogn)
        # return nums[-k]     


        #python's heap puts lowest at beginning
        heap = []
        for num in nums:
            heapq.heappush(heap, num)
            if len(heap) > k:
                heapq.heappop(heap)

        return heap[0]

#TC: O(n) for for loop, Heap operation is O(log k), since its inside the for loop, multiply ie O(nlogk)
#SC: O(k) since heap only stores k number of values

#this is slightly better than the first sort solution ie TC: O(nlogn)

#more comments
       #simple method is to nums.sort(), then return nums[-k] ie from the back Time: O(nlogn) space: O(1)

       #python's heap puts smallest value at top or beginning
        #iterate through the nums, push the vals onto the heap, then check if len(heap) > k
        # for num in nums:
        #     heapq.heappush(heap, num)

        # #if len(heap) > k, then pop the heap ie the smallest
        #     if len(heap) > k:
        #         heapq.heappop(heap)

        # return heap[0]
```
