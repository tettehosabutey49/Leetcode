You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) and (i, height[i]).

Find two lines that together with the x-axis form a container, such that the container contains the most water.

Return the maximum amount of water a container can store.

Notice that you may not slant the container.

 

Example 1:


Input: height = [1,8,6,2,5,4,8,3,7]
Output: 49
Explanation: The above vertical lines are represented by array [1,8,6,2,5,4,8,3,7]. In this case, the max area of water (blue section) the container can contain is 49.
Example 2:

Input: height = [1,1]
Output: 1

```python
class Solution(object):
    def maxArea(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        
        #track lp and rp and the width as well

        left = 0
        right = len(height)-1
        spaces = len(height)-1
        max_area = 0                        #used a list initially to append each area and selected the max. But optimized it
                                            #to meet the space O(1)

        
        while height and right > left:
            if height[left] < height[right]:            #when left value < rv, update current_area
                current_area = height[left]*spaces  
                max_area = max(max_area, current_area)      #get max of current_area and max_area
                left += 1

            elif height[left] >= height[right]:
                current_area = height[right]*spaces
                max_area = max(current_area, max_area)
                right -= 1
            spaces -= 1

        return max_area

```

Space Complexity: O(1)
Time complecity: O(n)

Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

 

Example 1:

Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
Explanation: 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.
The distinct triplets are [-1,0,1] and [-1,-1,2].
Notice that the order of the output and the order of the triplets does not matter.
Example 2:

Input: nums = [0,1,1]
Output: []
Explanation: The only possible triplet does not sum up to 0.
Example 3:

Input: nums = [0,0,0]
Output: [[0,0,0]]
Explanation: The only possible triplet sums up to 0.
 

Constraints:

3 <= nums.length <= 3000
-105 <= nums[i] <= 105

```python
class Solution(object):
    def threeSum(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        """
        [-4,-1,-1,0,1,2]
          a l         r
        """
        nums.sort()        
        ret = []
        

        for i in range(len(nums)):
            l = i+1
            r = len(nums)-1
            
            #skip any iteration where prev i and current i are the same and
            # i != 0 since if i = 0, it wouldnt have any prev element
            if nums[i] == nums[i-1] and i !=0:
                continue
            
            while l < r:
                total = nums[i] + nums[l] + nums[r]
                if total == 0:
                    ret.append([nums[i], nums[l],nums[r]])
                    l+=1
                    while l < r and nums[l] == nums[l-1]:
                        l+=1
                
                elif total < 0:
                    l+=1

                elif total > 0:
                    r-=1

        return ret



```

Time : sorting (O(nlogn)), outer loop and inner loop : O(n*n) final: O(n^2)

Space: O(n) if we count the output array or O(1) if we dont count the output array
