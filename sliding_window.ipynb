Given a string s, find the length of the longest substring without duplicate characters.

 

Example 1:

Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc", with the length of 3. Note that "bca" and "cab" are also correct answers.
Example 2:

Input: s = "bbbbb"
Output: 1
Explanation: The answer is "b", with the length of 1.

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        "abcabcbb"
        "curr is like the moving right pointer"
        last_seen_dict = {}
        start = 0 
        max_len = 0
        curr = 0
        
        
        for curr in range(len(s)):
            #if item in dict and index of item >= start, then update start index to be one after the initial item identified
            #as a duplicate
            if s[curr] in last_seen_dict and last_seen_dict[s[curr]] >= start:
                start = last_seen_dict[s[curr]] + 1

            #keep updating the dict with the last index of whatever item we are currently on in this iteration
            last_seen_dict[s[curr]] = curr       # update last seen position ie overwrite the last one

            # curr - start+1 to get the window length, max_len saves the prev window for us
            max_len = max(max_len, curr - start + 1)

        return max_len
```
