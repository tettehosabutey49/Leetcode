Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

 

Example 1:

Input: nums = [1,2,3,1]

Output: true

Explanation:

The element 1 occurs at the indices 0 and 3.

Example 2:

Input: nums = [1,2,3,4]

Output: false

Explanation:

All elements are distinct.

Example 3:

Input: nums = [1,1,1,3,3,4,3,2,4,2]

Output: true

 

```python
from collections import defaultdict
class Solution(object):
    def containsDuplicate(self, nums):
        """
        :type nums: List[int]
        :rtype: bool
        """

        # d = defaultdict(int)

        # for i, num in enumerate(nums):
        #     if num in d:
        #         return True 
        #     d[num] = i 
        # return False
#TC: O(n) we are just iterating through the nums
#SC: O(n) store n element in worst case scenario


#Using set (much cleaner)
        unique = set()
        for num in nums:
            if num in unique:
                return True 
            unique.add(num)
        return False
```
