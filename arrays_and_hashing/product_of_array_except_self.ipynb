Given an integer array nums, return an array answer such that answer[i] is equal to the product of all the elements of nums except nums[i].

The product of any prefix or suffix of nums is guaranteed to fit in a 32-bit integer.

You must write an algorithm that runs in O(n) time and without using the division operation.

 

Example 1:

Input: nums = [1,2,3,4]
Output: [24,12,8,6]
Example 2:

Input: nums = [-1,1,0,-3,3]
Output: [0,0,9,0,0]
 

Constraints:

2 <= nums.length <= 105
-30 <= nums[i] <= 30
The input is generated such that answer[i] is guaranteed to fit in a 32-bit integer.
 

Follow up: Can you solve the problem in O(1) extra space complexity? (The output array does not count as extra space for space complexity analysis.)

```python
class Solution(object):
    def productExceptSelf(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        #eg nums = [1,2,3,4,5]
        #lefts = [1,1,2,6,24]       nb: lefts[i] = product of all elements to the left of i
        #rights = [120,60,20,5,1]   nb: rights[i] = product of all elements to the right of i in the given array
        #answers = [120,60,40,30,24]
        #create array to store lefts
        left = [0]*len(nums)                #[0,0,0,0,0]
        
        #initialize the first element as 1
        left[0] = 1                         #[1,0,0,0,0]

        #loop through the rest of the lefts
        for i in range(1,len(nums)):
            left[i] = left[i-1]*nums[i-1]       #[1,1,2,6,24]

        #create array for right
        right = [0]*len(nums)
        #initialize the last element of right as 1
        right[-1] = 1                               #[0,0,0,0,1]

        #loop through the rest 
        for i in range(len(nums)-2, -1, -1):        #
            right[i]=right[i+1]*nums[i+1]           #[120,60,20,5,1]

        #create array for answers
        ans = [0]*len(nums)

        #loop through and multiply right and lefts
        for i in range(len(nums)):
            ans[i] = right[i]*left[i]               #[120,60,40,30,24]

        return ans                              #[120,60,20,30,24]
    
    ## Time complexity: O(n) ie right loop O(n), left loop O(n), final loop O(n) => O(3n) = O(n)
    ## Space complexity: O(n) ie left array O(n), right array O(n), answer array O(n) => O(3n) = O(n)

    ## FOllow up: can we do it in O(1) space compelxity?
    ## Yes we can simpler use the ans array to store the left products first then multiply by the right products on the fly

    

        
```

```python
## Here, we just use one list(ans) to store the left products and then multiply by the right products on the fly without storing them in a separate list. 
# We just use a variable(right) "instead of the right_list we used initially" to keep track of the right product as we iterate from right to left.
# this gives us O(1) in SC

class Solution(object):
    def productExceptSelf(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        #eg nums = [1,2,3,4,5]
        #ans = [1,1,2,6,24]       nb: ans[i] = product of all elements to the left of i
        #
        #store lefts in ans array
        ans = [0]*len(nums)
        
        #initialize the first element as 1
        ans[0] = 1                          #[1,0,0,0,0]   

        ## loop through the rest of the lefts(ans)
        for i in range(1,len(nums)):
            ans[i] = ans[i-1]*nums[i-1]
        
        right = 1                   #using just a variable to track the right product instead of a separate list

        #loop through from left to right while multiplying the various left and rights on the fly without storing in new array
        for i in range(len(nums)-1, -1, -1):
            ans[i] = right*ans[i]
            right = right * nums[i]

        return ans
    
    ## Time complexity: O(n) ie left loop O(n), right loop O(n) => O(2n) = O(n)
    ## Space complexity: O(1) ie ans array O(n) but does not count towards space complexity as per the problem statement(we need to return it)
```
