Given two strings s and t, return true if t is an anagram of s, and false otherwise.

 

Example 1:

Input: s = "anagram", t = "nagaram"

Output: true

Example 2:

Input: s = "rat", t = "car"

Output: false

 

```python
from collections import defaultdict
class Solution(object):
    def isAnagram(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: bool
        """
        if len(s) != len(t):
            return False
                                                # a b c d ....................z
        count = [0]*26                  #count = [0,0,0,0,0,0,00,,0,0,0,0,0,0,0]
        for i in range(len(s)):
            count[ord(s[i]) - ord('a')] += 1
            count[ord(t[i])-ord('a')] -= 1
    
        return True if count == [0]*26 else False

#TC: o(N) since we go through the range of the string once
#SC: O(1) since w estore constantly 26 characters irrespective of the size of the string.
```
