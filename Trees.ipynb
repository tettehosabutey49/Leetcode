# Pattern for Breath First Search


## All you need to know about most BFS tree problems
1. capture edge cases
2. trust the function to handle the children, left, right
3. combine the outputs from step two 
4. return it in whatever format expected

```python
# Maximum Depth
def maxDepth(root):
    # Step 1: Edge case - empty tree has depth 0
    if not root:
        return 0
    
    # Step 2: Trust the algorithm - get depth from children
    left_depth = maxDepth(root.left)
    right_depth = maxDepth(root.right)
    
    # Step 3: Combine results - take max depth and add 1 for current node
    return max(left_depth, right_depth) + 1


# Invert Tree
def invertTree(root):
    # Step 1: Edge case - empty tree returns None
    if not root:
        return None
    
    # Step 2: Trust the algorithm - invert both children
    left_inverted = invertTree(root.left)
    right_inverted = invertTree(root.right)
    
    # Step 3: Combine results - swap the inverted children
    root.left = right_inverted
    root.right = left_inverted
    return root


# Same Tree
def isSameTree(p, q):
    # Step 1: Edge cases
    if not p and not q:  # Both empty = same
        return True
    if not p or not q:   # One empty, one not = different
        return False
    if p.val != q.val:   # Values don't match = different
        return False
    
    # Step 2: Trust the algorithm - check if children are same
    left_same = isSameTree(p.left, q.left)
    right_same = isSameTree(p.right, q.right)
    
    # Step 3: Combine results - both subtrees must be same
    return left_same and right_same
```

Given the root of a binary tree, return its maximum depth.

A binary tree's maximum depth is the number of nodes along the longest path from the root node down to the farthest leaf node.
Input: root = [3,9,20,null,null,15,7]
Output: 3
Example 2:

Input: root = [1,null,2]
Output: 2

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def maxDepth(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        if root is None:
            return 0
        else:
            return 1 + max(self.maxDepth(root.left), self.maxDepth(root.right))
```

Given the roots of two binary trees p and q, write a function to check if they are the same or not.

Two binary trees are considered the same if they are structurally identical, and the nodes have the same value.
Input: p = [1,2,3], q = [1,2,3]
Output: true

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def isSameTree(self, p, q):
        """
        :type p: Optional[TreeNode]
        :type q: Optional[TreeNode]
        :rtype: bool
        """
        "Lets catch the edge cases first"
        if p is None and q is None:
            return True
        if (p is not None and q is not None and  p.val != q.val) or p is None or q is None:
            return False
        return self.isSameTree(p.left, q.left) and self.isSameTree(p.right, q.right)
```

Given the root of a binary tree, return the level order traversal of its nodes' values. (i.e., from left to right, level by level).

Input: root = [3,9,20,null,null,15,7]
Output: [[3],[9,20],[15,7]]
Example 2:

Input: root = [1]
Output: [[1]]
Example 3:

Input: root = []
Output: []

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def levelOrder(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: List[List[int]]
        """
        
        "result = [[level 0], [level 1,level1], [level 2, level 2]]"
        "NB at level 0, we create 1 sublist in our result, at level 1, our sublist will still be 1, so "
        "we need to create one more, and store all level 1 values there"

        results = []

        def dfs(node, level):
            
            #if root node is empty, dont do anything
            if not node:
                return

            #if level #0 == len(results) #0 then create subset for level 0 nodes/values
            if level == len(results):  
                results.append([])

            #store node value in level 0 sublist
            results[level].append(node.val)

            #handle left
            dfs(node.left, level+1)

            #handle right
            dfs(node.right, level+1)

        #call the function, starting from root, at level=0
        dfs(root,0)

        return results
        

```
