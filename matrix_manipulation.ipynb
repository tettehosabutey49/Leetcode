You are given an n x n 2D matrix representing an image, rotate the image by 90 degrees (clockwise).

You have to rotate the image in-place, which means you have to modify the input 2D matrix directly. DO NOT allocate another 2D matrix and do the rotation.

 

Example 1:


Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]
Output: [[7,4,1],[8,5,2],[9,6,3]]

```python
class Solution(object):
    def rotate(self, matrix):
        """
        :type matrix: List[List[int]]
        :rtype: None Do not return anything, modify matrix in-place instead.
        """
        # question
        # [[1,2,3],
        # [ 4,5,6],
        # [ 7,8,9]]

        # rotation
        #  [[7,4,1],
        #  [ 8,5,2],
        #  [ 9,6,3]]

        #  transpose matrix and compare it to the rotated result
        #  we see that the only difference is that our transposed one is reversed
        #  so we just reverse it back
        #  [[1, 4 , 7],
        #  [ 2, 5, 8 ],
        #  [ 3, 6, 9]]
        #   why? range(i+1, n), because we dont want to touch diagonals, and anything below the diagonals
        #   because we might end up doing the swap twice
        # j     0   1   2
        # i=0  [[1, 4 , 7],
        # i=1  [ 2, 5, 8 ],
        # i=2  [ 3, 6, 9]]

        n = len(matrix)                 #nb square matrix so n wroks for both row and col
        for i in range(n):
            for j in range(i+1, n):
                matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]         #swap elements of the matrix

        for i in range(n):                                            #iterate through the rows, and reverse the items
            matrix[i].reverse()

#Time complexity: O(n^2) nested_loops=O(n^2), matrix_swap=O(1), loop_through_rows = O(n) + reverse=O(n) ==> O(n^2)
#Space complexity: O(1) we are doing it in place
```

Given an m x n matrix, return all elements of the matrix in spiral order.

 

Example 1:


Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]
Output: [1,2,3,6,9,8,7,4,5]

```python
class Solution(object):
    def spiralOrder(self, matrix):
        """
        :type matrix: List[List[int]]
        :rtype: List[int]
        """
    #      question
    #          l            r
    #     top [1,  2,   3],
    #         [4,  5,   6],
    #         [7,  8,   9]
    #     dwn
        
        left = top = 0
        right = len(matrix[0])
        down = len(matrix)
        output = []
        
        while left < right and top < down:
            #append all items in top row(moving from left to right)
            for col in range(left, right):
                output.append(matrix[top][col])
            top += 1                                    #move down

            #append all items on rightmost column
            for row in range(top, down):
                output.append(matrix[row][right -1])
            right -= 1

            if left < right and top < down:
                #append all items on bottom (right to left)
                for col in range(right-1, left-1, -1):          #range(right-1....) bc we have tackled all right elements already(rightmost cols) in prev appending. 
                                                                #so we want the range(right - 1 ... to continue from val 8 instead of val 9 from the qtn
                    output.append(matrix[down-1][col])
                down -= 1

                #append all items on left(down to top)
                for row in range(down-1, top-1, -1):
                    output.append(matrix[row][left])

                left += 1

        return output

#Time complexity: O(mxn) we are visiting each element once
#Space complexity: O(1) we are using constant space for variables left, right,
```
